## Multiple Linear Regression — Predicting House Prices

### Objective
> Fit a multiple linear regression model to predict house prices based on size and number of bedrooms.

Step 1: Load the dataset from house_prices.csv

```python
import pandas as pd
data = pd.read_csv("./Data/Housing.csv",usecols=["price","area","bedrooms"])
data.sample(5)
```

Step 2:	Fit a linear regression model: Price = β0 + β1(Size) + β2(Bedrooms)

```python
# spliting the data into traning and testing data
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(data.iloc[:,[1,2]],data["price"],test_size=0.2,random_state=42)
print(f"xtrain:{xtrain.shape}\nxtest:{xtest.shape}")
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(xtrain,ytrain)
```

Step 3:	Display the regression coefficients and intercept

```python
print(f"regression coefficients: {model.coef_,}\nregression intercept: {model.intercept_}")
```

Step 4:	Predict the price for a house with Size = 1800 sq.ft and Bedrooms = 3

```python
import numpy as np
test=pd.DataFrame({"area":[1800],"bedrooms":[3]})
result = model.predict(test)
print(f"the price for a house with Size = 1800 sq.ft and Bedrooms = 3 will be Rs.{np.round(result,2)[0]}")
```

Step 5:	Evaluate the model using R² score on a held-out test set

```python
ypred = model.predict(xtest)

from sklearn.metrics import r2_score
score = r2_score(ytest,ypred)
print(f"The score of the model is {score*100:.3f}%")
```

Step 6:	Plot the regression plane in 3D (Size, Bedrooms vs. Price)

```python
import plotly.graph_objects as go
fig = go.Figure()

# Add scatter trace
fig.add_trace(go.Scatter3d(x=xtest["bedrooms"],y=xtest["area"],z=ytest, mode='markers', name='Markers'))

# Add line trace
fig.add_trace(go.Scatter3d(x=xtest["bedrooms"],y=xtest["area"],z=ypred, mode='lines', name='Lines'))

fig.update_layout(
    width=900,
    height=700,
    title="Resized 3D Overlay Plot"
)

fig.show()
```

### Conclusion
1. **Model Equation:** 
   The simple linear regression model was successfully trained with the equation:
   $$\text{Price} = 642086.408 + (389.2257 \times \text{area}) + (69567.8667 \times \text{bedrooms})$$
   * **Slopes ($\beta_1 \approx \$3.89225730e+02$ , \$6.95678667e+05$)
   * **Intercept ($\beta_0 \approx \$642086.408$)
