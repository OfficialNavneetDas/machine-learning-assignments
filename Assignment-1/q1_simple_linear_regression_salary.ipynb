## Simple Linear Regression — Salary vs. Years of Experience  

### Objective
> Fit a simple linear regression model to predict salary from years of experience.

### Step 1: Load the dataset from salary_data.csv.

```python
import pandas as pd
data = pd.read_csv("./data/Salary_dataset.csv",index_col=0)
data.head()
```

### Step 2: Split the data into training and test sets (e.g., 80/20)

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(data[["YearsExperience"]],data["Salary"],test_size=0.2,random_state=42)
print("x_train = ",x_train.shape)
print("x_test = ",x_test.shape)
```

### Step 3: Fit a linear regression model: y = β0 + β1x

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train,y_train)
```

### Step 4:	Display the intercept (β0) and slope (β1).

```python
print("β1 slope of the data",model.coef_[0])
print("β2 intercept of the data",model.intercept_)
```

### Step 5:	Predict the salary for a candidate with 7 years of experience.

```python
model.predict([[7]])
```

### Step 6: Evaluate the model using R² score and Mean Squared Error (MSE)

```python
# Generate prediction for the x_test
y_pred=model.predict(x_test)
display(y_pred)
```

```python
# Calculated matrics
from sklearn.metrics import mean_squared_error,r2_score

mse=mean_squared_error(y_test,y_pred)
r2 = r2_score(y_test,y_pred)

print(f"Mean squared error = {mse :.4f}")
print(f"r2 Score = ${r2 :.4f}")
```

### Step 7:	Plot the regression line over the scatter of training data.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

sns.scatterplot(x=x_train.values.ravel(),y=y_train.squeeze())
plt.plot(x_train,model.predict(x_train),c="r")
```
