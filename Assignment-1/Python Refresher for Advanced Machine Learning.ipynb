```python
# Define euclidean_distance(point1, point2) to compute the distance between two n-dimensional points.
import math
def euclidean_distance(x1, x2, y1, y2):
    return math.sqrt((x2-x1)**2+(y2-y1)**2)

print(euclidean_distance(0,6,4,2))
```

```python
# Define normalize(values) that scales a list of numbers to the range [0, 1].
def normalize(mylist):
    norm = [i/max(mylist) for i in mylist]
    return norm

li=[11,54,67,84,35,26,42,97,62,15,9]
print(normalize(li))
```

```python
# Perform matrix operations: transpose and matrix multiplication (dot product)
A = np.array([[1, 2, 3], 
              [4, 5, 6]])

B = np.array([[7, 8], 
              [9, 10], 
              [11, 12]])

# 2. Matrix Transpose
A_transposed = A.T  
print("Transpose of A:\n", A_transposed)

# 3. Matrix Multiplication (Dot Product) @ operator
result_operator = A @ B  
print("\nMatrix Multiplication (A @ B):\n", result_operator)

# Alternative standard syntax: np.dot()
result_dot = np.dot(A, B)
print(result_dot)
```

```python
# Generate random numbers using numpy.random, and compare the time taken for a vectorized operation versus an equivalent Python loop
import time
n = 1_000_000

# 1. Python Loop Approach
start_time = time.time()
python_list = []
for _ in range(n):
    python_list.append(np.random.rand())
loop_time = time.time() - start_time

# 2. Vectorized NumPy Approach
start_time = time.time()
numpy_arr = np.random.rand(n)
vector_time = time.time() - start_time

print(f"Python Loop Time: {loop_time:.4f} seconds")
print(f"Vectorized NumPy Time: {vector_time:.4f} seconds")
print(f"Speedup: {loop_time / vector_time:.1f}x faster")
```

# Question 5: Data Manipulation and Visualization with Pandas

```python
# Create a DataFrame with columns Name, Marks, Height and Weight
import pandas as pd
data = {
    'Name': ['navneet', 'prince', 'hemant',"vicky"],
    'Marks': [85, 90, 48,50],
    'Height': [165, 175, 160,140],
    'Weight': [55, 70, 62,51]
}
df = pd.DataFrame(data)
df
```

```python
# Filter students who scored above a chosen threshold, and add a derived Result column (Pass/Fail).
threshold = 60

# Syntax: np.where(condition, value_if_true, value_if_false)
df["Result"] = np.where(df["Marks"] >= threshold, "Pass", "Fail")

passed_students = df[df["Result"] == "Pass"]
print("--- Original DataFrame with Result Column ---")
print(df)
print("\n--- Filtered DataFrame (Passed Students Only) ---")
print(passed_students)
```

```python
# Group the data by Result and compute summary statistics for each group.
z = df.groupby("Result")
z.get_group("Pass").describe()
```

```python
# Group the data by Result and compute summary statistics for each group.
z = df.groupby("Result")
z.get_group("Fail").describe()
```

```python
# Create a line plot, a bar plot and a histogram to visualize different aspects of the data.
import matplotlib.pyplot as plt

# Set up a 1x3 grid for side-by-side plots
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# ----------------------------------------------------
# 1. Line Plot: Height and Weight Trends across Students
# ----------------------------------------------------
df.plot(
    kind="line",
    x="Name",
    y=["Height", "Weight"],
    marker="o",
    ax=axes[0],
    color=["#3498db", "#e74c3c"],
)
axes[0].set_title("Height & Weight by Student")
axes[0].set_ylabel("Measurements (cm / kg)")
axes[0].grid(True, linestyle="--", alpha=0.6)

# ----------------------------------------------------
# 2. Bar Plot: Average Marks by Result Group
# ----------------------------------------------------
avg_marks = df.groupby("Result")["Marks"].mean()
avg_marks.plot(kind="bar", ax=axes[1], color=["#e74c3c", "#2ecc71"])
axes[1].set_title("Average Marks: Pass vs Fail")
axes[1].set_ylabel("Average Marks")
axes[1].set_xticklabels(axes[1].get_xticklabels(), rotation=0)
axes[1].grid(axis="y", linestyle="--", alpha=0.6)

# ----------------------------------------------------
# 3. Histogram: Distribution of Student Marks
# ----------------------------------------------------
df["Marks"].plot(
    kind="hist", bins=5, ax=axes[2], color="#9b59b6", edgecolor="black"
)
axes[2].set_title("Distribution of Marks")
axes[2].set_xlabel("Marks Range")
axes[2].grid(axis="y", linestyle="--", alpha=0.6)

# Adjust layout and display charts
plt.tight_layout()
plt.show()
```

### Discuss why normalization is generally necessary before using distance-based or gradient-based ML algorithms
## Algorithms like **K-Nearest Neighbors (KNN)**, **K-Means clustering**, and **Support Vector Machines (SVM)** calculate the distance (such as Euclidean distance) between data points.
> **Preventing dominance:** If one feature has a large numerical range (like annual salary in dollars from $30,000 to $150,000) and another has a small range (like age from 20 to 80), the larger numbers will completely control the distance calculation.  
> **Ensuring fair contribution:** Normalizing forces all features into the same range (typically 0 to 1), allowing every feature to contribute equally to the final similarity or distance score
